```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error,root_mean_squared_error,r2_score
```

```python
df=pd.read_csv("Advertising.csv")
```

```python
sns.heatmap(df.corr(),annot=True)
```

```python
df.drop('Unnamed: 0',axis=1,inplace=True)
```

```python
df.head()
```

```python
# Separate input and output features
```

```python
X=df[['TV']] #independent variable
Y=df[["Sales"]] #dependent variable
```

```python
# perform splitting of training and testing data
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=42)
```

```python
# Create an object of LinearRegression
```

```python
lr=LinearRegression()
```

```python
# this object is going to call OLS(ordinary least square function) which is used to 
# give best theta1 and theta0 value
```

```python
lr.fit(X_train,Y_train)
```

```python
lr.intercept_
```

```python
lr.coef_
```

```python
# Now next we have to calculate predicted value on training data
```

```python
Y_pred=lr.predict(X_train)
```

```python
plt.figure()
plt.scatter(X_train,Y_train)
plt.plot(X_train,Y_pred)
plt.show()
```

```python
# Now perform prediction on testing data
```

```python
y_pred=lr.predict(X_test)
```

```python
# Model Evaluation
```

```python
mse=mean_squared_error(y_pred,Y_test)
rmse=root_mean_squared_error(y_pred,Y_test)
mae=mean_absolute_error(y_pred,Y_test)
```

```python
print(mse)
print(rmse)
print(mae)
```
